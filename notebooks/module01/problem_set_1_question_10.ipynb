## Question 10

The shorthand $S$ for spam, $H$ for ham, $F$ for free, and $V$ for Viagra is used.

Given the information:

$$
\begin{align*}
    P(S) &= 0.6 \\
    P(H) &= 1 - P(S) = 0.4 \\
    P(S \mid F) &= 0.75 \\
    P(S \mid V) &= 0.9 \\
    P(H \mid F) &= 1 - P(S \mid F) = 0.25 \\
    P(H \mid V) &= 1 - P(S \mid V) = 0.1
\end{align*}
$$

It is also given that $F \perp V \mid S$, meaning that $P(F \mid V \cap S) = P(F \mid S)$ and $P(V \mid F \cap S) = P(V \mid S)$.

With this information, it is possible to answer the question $P(S \mid V \cap F)$.

First, using Bayes Rule[1], this probability is equivalent to:

$$
\begin{align*}
    P(S \mid V \cap F) &= \frac{P(V \cap F \mid S) P(S)}{P(V \cap F)}
\end{align*}
$$

By the Conditionalized Chain Rule[1], combined with the fact that $F \perp V \mid S$, the first part of the numerator is equivalent to:

$$
\begin{align*}
    P(V \cap F \mid S) &= P(V \mid F \cap S)P(F \mid S) \\
    &= P(V \mid S)P(F \mid S)
\end{align*}
$$

The denominator can be expressed with the Law of Total Probability[1], also combined combined with the fact that $F \perp V \mid S$ (as well as the assumption since $H$ is just the binary compliment of $S$, $F \perp V \mid \neg S \to F \perp V \mid H$), as:

$$
\begin{align*}
    P(V \cap F) &= P(V \cap F | S)P(S) + P(V \cap F|H)P(H) \\
        &= P(V \mid S)P(F \mid S)P(S) + P(V \mid H)P(F \mid H)P(H)
\end{align*}
$$



This means the original expression can now be stated as:

$$
\begin{align*}
    P(S \mid V \cap F) &= \frac{P(V \cap F \mid S) P(S)}{P(V \cap F)} \\
        &= \frac{P(V \mid S) P(F \mid S) P(S)}{P(V \mid S)P(F \mid S)P(S) + P(V \mid H)P(F \mid H)P(H)}
\end{align*}
$$

The value for $P(V \mid S)$ can be stated using Bayes Rule, combined with our known probability values as:

$$
\begin{align*}
    P(V \mid S) &= \frac{P(S \mid V)P(V)}{P(S)} \\
        &= \frac{S \mid V}{P(S)} P(V) \\
        &= \frac{0.9}{0.6} P(V) \\
        &= 1.5 P(V)
\end{align*}
$$

Similarly for $P(F \mid S)$:

$$
\begin{align*}
    P(F \mid S) &= \frac{P(S \mid F)P(F)}{P(S)} \\
        &= \frac{S \mid F}{P(S)} P(F) \\
        &= \frac{0.75}{0.6} P(F) \\
        &= 1.25 P(F)
\end{align*}
$$

The values for $P(V \mid H)$ and $P(F \mid H)$ can be found my using the Law of Total Probability, combined with these conditional probabilities:

For $P(V \mid H)$:

$$
\begin{align*}
    P(V) &= P(V \mid S)P(S) + P(V \mid H)P(H) \\
    P(V) &= (1.5 P(V))(0.6) + P(V \mid H)(0.4) \\
    P(V) - (1.5 P(V))(0.6) &= P(V \mid H)(0.4) \\
    \frac{P(V) - (1.5 P(V))(0.6)}{0.4} &= P(V \mid H) \\
    P(V \mid H) &= \frac{P(V) - 0.9P(V)}{0.4} \\
    P(V \mid H) &= \frac{1 - 0.9}{0.4} P(V) \\
    P(V \mid H) &= 0.25 P(V) \\
\end{align*}
$$

For $P(F \mid H)$:

$$
\begin{align*}
    P(F) &= P(F \mid S)P(S) + P(F \mid H)P(H) \\
    P(F) &= (1.25 P(F))(0.6) + P(F \mid H)(0.4) \\
    P(F) - (1.25 P(F))(0.6) &= P(F \mid H)(0.4) \\
    \frac{P(F) - (1.25 P(F))(0.6)}{0.4} &= P(F \mid H) \\
    P(F \mid H) &= \frac{P(F) - 0.75P(F)}{0.4} \\
    P(F \mid H) &= \frac{1 - 0.75}{0.4} P(F) \\
    P(F \mid H) &= 0.625 P(F) \\
\end{align*}
$$

Returning to the original expression:

$$
\begin{align*}
    P(S \mid V \cap F) &= \frac{P(V \cap F \mid S) P(S)}{P(V \cap F)} \\
        &= \frac{P(V \mid S) P(F \mid S) P(S)}{P(V \mid S)P(F \mid S)P(S) + P(V \mid H)P(F \mid H)P(H)} \\
        &= \frac{(1.5)P(V) (1.25)P(F) (0.6)}{(1.5)P(V)(1.25)P(F)(0.6) + (0.25)P(V)(0.625)P(F)(0.4)} \\
        &= \frac{1.125 P(V) P(F)}{1.875 P(V) P(F)} \\
        &= \frac{1.125}{1.1875} \\
        &\approx \boxed{0.9474}
\end{align*}
$$
